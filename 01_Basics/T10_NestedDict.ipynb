```python
employee = {"name": "Amit", 
"department": "Engineering",
"skills": {"language": "Python", "database": "PostgreSQL", "cloud": "AWS"},
"salary": 80000}
print(employee)
```

```python
#Print employee name
print("Employee Name:" ,employee["name"])
```

```python
#Print department
print("Employee department:" ,employee["department"])
```

```python
#Print complete skills dictionary
print("Employee Skills list:" ,employee["skills"])
```

```python
#Print programming language
print("Employee Programming Skills :" ,employee["skills"]["language"])
```

```python
#Print database
print("Employee Database Skills :" ,employee["skills"]["database"])
```

```python
#Print cloud technology
print("Employee Cloud Skills :" ,employee["skills"]["cloud"])
```

```python
#Change "Python" to "Python + JavaScript"
employee["skills"]["language"] = "Python + JavaScript"
print("Employee Programming Skills :" ,employee["skills"]["language"])
```

```python
#Change salary
print("Old Salary: ", employee["salary"])
employee["salary"]= 100000
print("New Salary: ", employee["salary"])
```

```python
#Add "experience": 3
employee["experience"] = 3
print("Employee Experience :", employee["experience"])
```

```python
#Add another skill under the skills dictionary
employee["skills"]["framework"] = "React"
print("Updated Skills:", employee["skills"])
```

```python
#Print complete updated employee dictionary
print("Updated Employee:", employee)
```
