```python
###Remove Duplicate Data Using Sets###
```

```python
#Oringinal List
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]
```

```python
#Print the original list
print("Original list:", numbers)
```

```python
#Convert the list into a set
unique_numbers = set(numbers)
print("Set:", unique_numbers)
print("type of numbers:", type(numbers))
print("type of coverted numbers:", type(unique_numbers))
```

```python
# Convert the set back into a list
unique_numbers_list = list(unique_numbers)
print("Unique numbers list:", unique_numbers_list)
print("type of coverted numbers:",type(unique_numbers))
print("type of re-coverted numbers:",type(unique_numbers_list))
```

```python
#Print number of original elements
print("Number of original elements:", len(numbers))
```

```python
#Print number of unique elements
print("Number of unique elements:", len(unique_numbers))
```

```python
#Another example with duplicate student names
students = ["Rahul", "Priya", "Aman", "Rahul", "Neha", "Priya", "Aman"]
print("Original students:", students)
print(type(students))
```

```python
#Convert list to set
unique_students = set(students)
print("Unique students:", unique_students)
print(type(unique_students))
```

```python
#Convert set back to list
unique_students_list = list(unique_students)
print("Unique student list:", unique_students_list)
print(type(unique_students_list))
```

```python
#Count original students
print("Original count:", len(students))
```

```python
#Count unique students
print("Unique count:", len(unique_students))
```
