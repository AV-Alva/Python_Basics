```python
#creating a string variable
message = "   Welcome To Python Programming Class   "
print(message)
```

```python
#remove extra spaces from the string
message1 = message.strip()
print(message1)
```

```python
#Convert everything to lowercase
message2 = message1.lower()
print(message2)
```

```python
#Convert everything to uppercase
message3 = message1.upper()
print(message3)
```

```python
#Convert to title case
message4 = message1.title()
print(message4)
```

```python
#Replace "Python" with "Advanced Python"
message5 = message1.replace("Python", "Advanced Python")
print(message5)
```

```python
#Check whether the string starts with "Welcome"
message6 = message1.startswith("Welcome")
print(message6)
```

```python
#Check whether it ends with "Class"
message7 = message1.endswith("Class")
print(message7)
```

```python
#Count occurrences of "o"
message8 = message1.count("o")
print(message8)
```

```python
#Find the position of "Programming"
message9 = message1.find("Programming")
print(message9)
```

```python
#Split the sentence into words  
message10 = message1.split()
print(message10)
```

```python
#Explanation of the methods used in the code:
# strip() - removes leading and trailing whitespace
# lower() - converts all characters in the string to lowercase
# upper() - converts all characters in the string to uppercase
# title() - converts the first character of each word to uppercase
# replace() - replaces a specified phrase with another specified phrase
# startswith() - checks if the string starts with a specified phrase
# endswith() - checks if the string ends with a specified phrase
# find() - Finds the index of the first occurrence of specified text
# count() - counts the number of occurrences of a specified value
# split() - splits the string into a list where each word is a list item
```
