```python
student = {"name": "Rahul",
           "age": 22,
           "course": "Python",
           "city": "Bangalore",
           "marks": 88}
```

```python
#Print the complete dictionary
print("Student details: ", student)
```

```python
#Print the student's name
print("Student Name : ", student["name"])
```

```python
#Print their course
print("Course : ", student["course"])
```

```python
#Print all keys
print("All Keys : ", student.keys())
```

```python
#Print all values
print("All Values : ", student.values())
```

```python
#Print all key-value pairs
print("Key-Value Pairs : ", student.items())
```

```python
#Change marks from 88 to 92
print("Old marks : ", student["marks"])
student["marks"] = 92
print("New marks : ", student["marks"])
```

```python
#Add "email"
student["email"]="abc.gmail.com"
print("Updated Student details", student)
```

```python
#Add "phone"
student["phone"]="39137912379"
print("Updated Student details", student)
```

```python
#Remove "city"
student.pop("city")
print("Updated Student details", student)
```

```python
#Use get() to retrieve "name"
print("Name", student.get("name"))
```

```python
#Create a copy of the dictionary
student_copy = student.copy()
print("Copied Student details", student_copy)
```

```python
#Dictionary Methods
#keys() → Returns all keys in the dictionary.
#values() → Returns all values in the dictionary.
#items() → Returns all key-value pairs.
#get() → Returns the value for a given key without raising an error if the key is missing.
#update() → Adds new key-value pairs or updates existing ones.
#pop() → Removes a specified key and returns its value.
#copy() → Creates a shallow copy of the dictionary.
```
