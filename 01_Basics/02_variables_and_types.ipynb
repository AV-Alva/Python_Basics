# Variables and Data Types

Understanding Python's fundamental data types

```python
# Variables and Assignment
age = 25
name = 'Alice'
height = 5.8
is_student = True

print(f'Name: {name}')
print(f'Age: {age}')
print(f'Height: {height}')
print(f'Is Student: {is_student}')
```

```python
# Check data types
print(type(age))
print(type(name))
print(type(height))
print(type(is_student))
```

```python
# Type conversion
age_string = str(age)
print(f'Age as string: {age_string}')
print(type(age_string))
```

## Data Types in Python
- **int**: Integer numbers (25, -10, 0)
- **float**: Decimal numbers (5.8, 3.14)
- **str**: Text strings ('Alice', "Hello")
- **bool**: Boolean values (True, False)
