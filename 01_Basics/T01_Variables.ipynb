```python
#Get all details from user 

Name = input("Enter your name: ")
Age = int(input("Enter your age: "))
City = input("Enter your city: ")
Company_Name = input("Enter your company name: ")
Topic = input("Enter your current learning topic: ")
Experience = int(input("Enter your years of experience/study: "))
Like = bool(input("Do you like Python? (true/false): "))
Expected_Salary = float(input("Enter your expected salary after learning Python: "))
```

```python
#print the details received from user

print("Details received from user:")
print(Name)
print(Age)
print(City)
print(Company_Name)
print(Topic)
print(Experience)
print(Like)
print(Expected_Salary)
```

```python
#print data type of every variable

print("Data types of the variables:")
print("Data type of Name:", type(Name))
print("Data type of Age:", type(Age))
print("Data type of City:", type(City))
print("Data type of Company Name:", type(Company_Name))
print("Data type of Learning Topic:", type(Topic))
print("Data type of Experience:", type(Experience))
print("Data type of Liking Python:", type(Like))
print("Data type of Expected Salary:", type(Expected_Salary))
```
