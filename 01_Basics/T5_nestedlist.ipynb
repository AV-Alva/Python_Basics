```python
students = [["Rahul", 21, "Python"],
            ["Priya", 22, "Data Science"],
            ["Aman", 20, "Machine Learning"]]
```

```python
#Print the complete list
print(students)
```

```python
#Print Rahul's name
print(students[0][0])
```

```python
#Print Priya's age
print(students[1][1])
```

```python
#Print Aman's course
print(students[2][2])
```

```python
#Print the complete record of Priya
print(students[1])
```

```python
#Change Rahul's course to "AI"
students[0][2] = "AI"
print(students[0])
```

```python
#Add another student record manually
students.append(["Amrutha", 35, "Generative AI"])
print(students[3])
```

```python
#Print the updated nested list
print(students)
```
