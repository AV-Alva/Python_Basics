```python
students = [["Rahul", 21, "Python"],
            ["Priya", 22, "Data Science"],
            ["Aman", 20, "Machine Learning"]]
```

```python
#Print the complete list
print(students)
```

```python
#Print Rahul's name
print(students[0][0])
```

```python
#Print Priya's age
print(students[1][1])
```

```python
#Print Aman's course
print(students[2][2])
```

```python
#Print the complete record of Priya
print(students[1])
```

```python
#Change Rahul's course to "AI"
students[0][2] = "AI"
print(students[0])
```

```python
#Add another student record manually
students.append(["Amrutha", 35, "Generative AI"])
print(students[3])
```

```python
#Print the updated nested list
print(students)
```

```python
#Indexing in Nested Lists
#A nested list is a list that contains one or more lists inside it.

#example:

#students = [
#    ["Rahul", 21, "Python"],
#    ["Priya", 22, "Data Science"],
#    ["Aman", 20, "Machine Learning"]
#]

#Python indexing starts from 0.

#Outer List Indexing

#The first index selects an entire inner list (student record).

#students[0] → ["Rahul", 21, "Python"]
#students[1] → ["Priya", 22, "Data Science"]
#students[2] → ["Aman", 20, "Machine Learning"]

#Inner List Indexing

#Each student record is itself a list:

#                [0]       [1]        [2]
#students[0] → ["Rahul",    21,      "Python"]
#students[1] → ["Priya",    22,      "Data Science"]
#students[2] → ["Aman",     20,      "Machine Learning"]

#To access an individual value, we use two indexes:

#students[outer_index][inner_index]

#The first index selects the inner list, and the second index selects an item from that inner list.
```
