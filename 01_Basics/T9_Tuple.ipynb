```python
technologies = ("Python", "Java", "Python", "C++", "JavaScript", "Python")
```

```python
#Print the tuple
print(technologies)
```

```python
#Print its type
print(type(technologies))
```

```python
#Print the first item
print("First item:", technologies[0])
```

```python
#Print the last item
print("Last item:", technologies[-1])
```

```python
#Slice the tuple
print(technologies[0:6])
```

```python
#Count "Python"
print("Python count:", technologies.count("Python"))
```

```python
#Find the index of "C++"
print("Index of C++:", technologies.index("C++"))
```

```python
#Find the tuple length
print("Tuple length:", len(technologies))
```

```python
#Convert tuple into a list
technology_list = list(technologies)
print(technology_list)
print(type(technology_list))
```

```python
#Add "Go" after converting it into a list
technology_list.append("Go")
print("After adding Go:", technology_list)
print(type(technology_list))
```

```python
#Convert it back into a tuple
Con_technologies = tuple(technology_list)
print(Con_technologies)
print(type(Con_technologies))
```
