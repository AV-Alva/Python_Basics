```python
products = [{"name": "Laptop","price": 70000,"brand": "Dell"},
            {"name": "Phone","price": 40000,"brand": "Samsung"},
            {"name": "Tablet","price": 30000,"brand": "Apple"}]
```

```python
#Print all products
print("All Products:", products)
```

```python
#Print first product
print("First Product:", products[0])
```

```python
#Print second product's price
print("Second Product's Price:", products[1]["price"])
```

```python
#Print third product's brand
print("Third Product's Brand:", products[2]["brand"])
```

```python
#Change first product's price
print("first product's price from", products[0]["price"], "to 75000")
products[0]["price"] = 75000
```

```python
#Add "rating" to the second product
products[1]["rating"] = 4.5
print("Second Product:", products[1])
```

```python
#Add another product manually
products.append({
    "name": "Smartwatch",
    "price": 20000,
    "brand": "Fitbit",  
})
```

```python
#Print final dataset
print("Final Dataset:", products)
```

```python
#### Understanding 

#List → `products` is a list containing multiple product records.
#Dictionary → Each product is stored as a dictionary.
#Indexing → Used to select a specific product. Example: `products[0]`
#Keys → Identify the fields inside a product, such as `"name"`, `"price"` and `"brand"`.
#Values → The actual data stored against each key, such as `"Laptop"`, `70000` and `"Dell"`.

#To access a specific value:
#products[0]["name"]
#Here, `[0]` selects the **first product**, and `["name"]` selects its **name**.
```
