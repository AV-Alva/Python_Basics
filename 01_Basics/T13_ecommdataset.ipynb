```python
products = [{"name": "Laptop","price": 70000,"brand": "Dell"},
            {"name": "Phone","price": 40000,"brand": "Samsung"},
            {"name": "Tablet","price": 30000,"brand": "Apple"}]
```

```python
#Print all products
print("All Products:", products)
```

```python
#Print first product
print("First Product:", products[0])
```

```python
#Print second product's price
print("Second Product's Price:", products[1]["price"])
```

```python
#Print third product's brand
print("Third Product's Brand:", products[2]["brand"])
```

```python
#Change first product's price
print("first product's price from", products[0]["price"], "to 75000")
products[0]["price"] = 75000
```

```python
#Add "rating" to the second product
products[1]["rating"] = 4.5
print("Second Product:", products[1])
```

```python
#Add another product manually
products.append({
    "name": "Smartwatch",
    "price": 20000,
    "brand": "Fitbit",  
})
```

```python
#Print final dataset
print("Final Dataset:", products)
```
