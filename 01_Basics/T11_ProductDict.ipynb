```python
laptop = {"brand": "Dell",
"model": "XPS 15",
"price": 120000,
"ram": "16GB",
"storage": "512GB SSD",
"available": True}
```

```python
#Display the product
print("Product :", laptop)
```

```python
#Print brand
print("Brand :", laptop["brand"])
```

```python
#Print model
print("model :", laptop["model"])
```

```python
#Print price
print("Price : ", laptop["price"])
```

```python
#Change price
print("Old price :", laptop["price"])
laptop["price"] = 150000
print("New price :", laptop["price"])
```

```python
#Add processor information
laptop["Processor"]= "Intel Core Ultra 5"
print("Updated laptop details: ", laptop)
```

```python
#Add GPU information
laptop["GPU"] = "Intel Iris Xe"
print("Updated laptop details: ", laptop)
```

```python
#Change RAM to "32GB"
print("Old RAM :", laptop["ram"])
laptop["ram"] = "32GB"
print("New RAM :", laptop["ram"])
```

```python
#Remove "available"
laptop.pop("available")
print("After removing available:", laptop)
```

```python
#Print all keys
print("Keys :", laptop.keys())
```

```python
#Print all values
print("Values :", laptop.values())
```

```python
#Print all items
print("All items: ", laptop.items())
```
