```python
###Student Information String Challenge###
```

```python
student = "python programming for data science"
```

```python
#Print the complete string#
print(student)
```

```python
#Print the first character#
print(student[0])
```

```python
#Print the last character#
print(student[-1])
```

```python
#Print the first 6 characters#
print(student[0:6])
```

```python
#Print the last 7 characters#
print(student[-7:])
```

```python
#Reverse the string using slicing#
reversed_student = student[::-1]
print(reversed_student)
```

```python
#Convert it to uppercase#   
student1 = student.upper()
print(student1)
```

```python
#Convert it to lowercase#
student2 = student.lower()
print(student2)
```

```python
#Convert it to title case#
student3 = student.title()
print(student3)
```

```python
#Count how many times "a" appears#
print(student.count('a'))
```

```python
#Find the position of "programming"#
student.find("programming")
```

```python
#Replace "data science" with "artificial intelligence"
student4 = student.replace("data science","artificial intelligence")
print(student4)
```

```python
#Split the string into individual words#
student5 = student.split()
print(student5)
```
