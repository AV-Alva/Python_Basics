```python
###Python Data Type Laboratory###
```

```python
#Integer - Int#
age = 35
year = 2026
print(type(age))
print(type(year))
```

```python
# Float - float#
salary = 50000.50
temperature = 28.5
print(type(salary))
print(type(temperature))
```

```python
# String - str
name = "Amrutha"
course = "Python Programming"
print(type(name))
print(type(course))
```

```python
# Boolean - bool
is_student = True
is_completed = False
print(type(is_student))
print(type(is_completed))
```

```python
# List - list
fruits = ["apple", "banana", "mango"]
marks = [85, 90, 78, 95]
data_l = [100, 45.5, "Python", True]
print(type(fruits))
print(type(marks))
print(type(data_l))
```

```python
# Tuple - tuple
coordinates = (10, 20)
colors = ("red", "green", "blue")
data_t = (100, 45.5, "Python", True)
print(type(coordinates))
print(type(colors))
print(type(data_t))
```

```python
# Set - set
numbers = {1, 2, 3, 4}
languages = {"Python", "Java", "SQL"}
data_s = {100, 45.5, "Python"}
print(type(numbers))
print(type(languages))
print(type(data_s))
```

```python
#Tuple inside set#
data_ts = {100, "Python", (1, 2, 3)}
print(type(data_ts))
```

```python
#unique values only in set#
numbers = {1, 2, 3, 4, 5, 1, 2, 2, 9, 5,6}
languages = {"Python", "Java", "SQL", "python", "SQL", "java"}
print(numbers)
print(languages)
```

```python
# Dictionary - dict#
student = {"name": "Amrutha", "age": 35, "course": "Python"}
employee = {"id": 101, "role": "Manager"}
print(type(student))
print(type(employee))
```

```python
###Type Conversion###
```

```python
#String to Integer#
value1 = int("100")
print(value1)
print(type(value1))
#Here string "100" is converted to integer 100 using int() function.
```

```python
#String to Float#
value2 = float("45.67")
print(value2)
print(type(value2))
#Here string "45.67" is converted to float 45.67 using float() function.
```

```python
#Integer to String#
value3 = str(500)
print(value3)
print(type(value3))
#Here integer 500 is converted to string "500" using str() function.
```

```python
#Integer to Boolean#
value4 = bool(1)
print(value4)
print(type(value4))
#Here integer 1 is converted to boolean True using bool() function.
```

```python
#Tuple to List#
value5 = list((1, 2, 3))
print(value5)
print(type(value5))
#Here tuple (1, 2, 3) is converted to list [1, 2, 3] using list() function.
```

```python
#List to Tuple#
value6 = tuple([1, 2, 3])
print(value6)
print(type(value6))
#Here list [1, 2, 3] is converted to tuple (1, 2, 3) using tuple() function.
```

```python
#List to Set#
value7 = set([1, 2, 2, 3])
print(value7)
print(type(value7))
#Here list [1, 2, 2, 3] is converted to set {1, 2, 3} using set() function. Duplicates are removed in the process.
```
