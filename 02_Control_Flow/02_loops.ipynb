# Loops

Repeat code efficiently with for and while loops

```python
# For loop with range
for i in range(5):
    print(f'Iteration {i}')
```

```python
# For loop through a list
fruits = ['apple', 'banana', 'cherry']

for fruit in fruits:
    print(f'I like {fruit}')
```

```python
# While loop
count = 0
while count < 5:
    print(f'Count: {count}')
    count += 1
```

## Loop Control
- `break`: Exit the loop
- `continue`: Skip to next iteration
