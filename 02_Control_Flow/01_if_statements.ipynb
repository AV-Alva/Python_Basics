# Control Flow - If Statements

Learn how to make decisions in Python

```python
# Simple if statement
age = 18

if age >= 18:
    print('You are an adult')
else:
    print('You are a minor')
```

```python
# If-elif-else
score = 85

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'F'

print(f'Your grade: {grade}')
```

## Comparison Operators
- `==` : Equal to
- `!=` : Not equal to
- `>` : Greater than
- `<` : Less than
- `>=` : Greater than or equal
- `<=` : Less than or equal
