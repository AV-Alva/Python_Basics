# Lists

Working with Python's most versatile data structure

```python
# Creating lists
fruits = ['apple', 'banana', 'cherry']
numbers = [1, 2, 3, 4, 5]
mixed = [1, 'apple', 3.14, True]

print(fruits)
print(numbers)
print(mixed)
```

```python
# Accessing elements
print(fruits[0])  # First element
print(fruits[-1])  # Last element
print(fruits[0:2])  # Slice
```

```python
# List methods
fruits.append('orange')
print(fruits)

fruits.remove('banana')
print(fruits)

print(len(fruits))
```

## Common List Methods
- `append()`: Add item to end
- `remove()`: Remove item by value
- `pop()`: Remove item by index
- `sort()`: Sort the list
- `len()`: Get list length
